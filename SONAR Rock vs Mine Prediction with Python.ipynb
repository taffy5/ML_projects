```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
```

## Data collection & Processing

```python
# loading the dataset to a pandas DataFrame

sonar_data = pd.read_csv("sonar_data.csv", header = None)
```

```python
sonar_data.head(5)
# 61 columns cause in python the indexing starts from 0
```

```python
sonar_data.shape
```

```python
sonar_data.describe()
# describe only consider integer values
```

```python
sonar_data[60].value_counts()

# M - Mine
# R - Rock
```

```python
sonar_data.groupby(60).mean()
```

```python
# seperate data and labels
X = sonar_data.drop(columns=60, axis = 1)
Y = sonar_data[60]
```

```python
print(X)
print(Y)
```

## Train test data

```python
X_train, X_test, Y_train, Y_test = train_test_split(X,Y, test_size = 0.1, stratify=Y, random_state=1)

# stratify -> so that the split has both M and R values
# random_state -> so that the split doesnt shuffle everytime.
```

```python
print(X.shape, X_train.shape, X_test.shape)
```

```python
print(Y.shape, Y_train.shape, Y_test.shape)
```

```python
print(X_train)
print(Y_train)
```

## Model Training -> logistic regression

```python
# training the logistic regression model with training data

model = LogisticRegression()
model.fit(X_train, Y_train)
```

## Model Evaluation

```python
# accuracy on the training data

# any accuracy greater than 70% is good
X_train_prediction = model.predict(X_train)
training_data_accuracy = accuracy_score(X_train_prediction, Y_train)
```

```python
print(training_data_accuracy)
```

```python
# accuracy on the test data

# any accuracy greater than 70% is good
X_test_prediction = model.predict(X_test)
test_data_accuracy = accuracy_score(X_test_prediction, Y_test)
```

```python
print(test_data_accuracy)
```
